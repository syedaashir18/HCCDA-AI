```python

total = 0
count = 0
largest = 0
smallest = 0

while True:
    num = float(input("Enter a number (0 to stop): "))
    if num == 0:
        break
    total = total + num
    count = count + 1
    if count == 1:
        largest = num
        smallest = num
    else:
        if num > largest:
            largest = num
        if num < smallest:
            smallest = num

if count > 0:
    print("Total:", total)
    print("Average:", total / count)
    print("Maximum:", largest)
    print("Minimum:", smallest)
else:
    print("No numbers entered")
```

```python
password = input("Enter a password: ")

has_upper = False
has_lower = False
has_digit = False
has_special = False

for ch in password:
    if ch.isupper():
        has_upper = True
    elif ch.islower():
        has_lower = True
    elif ch.isdigit():
        has_digit = True
    else:
        has_special = True

if len(password) >= 8 and has_upper and has_lower and has_digit and has_special:
    print("Strong password")
else:
    print("Weak password")
```

```python
paragraph = input("Enter a paragraph: ")

vowels = 0
spaces = 0
digits = 0

for ch in paragraph.lower():
    if ch in "aeiou":
        vowels = vowels + 1
    if ch == " ":
        spaces = spaces + 1
    if ch.isdigit():
        digits = digits + 1

print("Characters:", len(paragraph))
print("Words:", len(paragraph.split()))
print("Vowels:", vowels)
print("Spaces:", spaces)
print("Digits:", digits)
```
